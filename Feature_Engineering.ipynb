```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('EDA_done.csv')
df.head(10).T
```

```python
print(df.shape)
print(df.columns)
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=['Blood Pressure (mmHg)'])
```

```python
df = pd.get_dummies(df,columns=["Sex"] , dtype=int)
```

```python
df.head(10).T
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories=[['Low','High','Moderate'],['LOW', 'INTERMEDIARY', 'HIGH'],['Normal', 'Elevated', 'Hypertension Stage 1', 'Hypertension Stage 2']],dtype=int)

col = ['Physical Activity Level','CVD Risk Level','Blood Pressure Category']

df[col] = oe.fit_transform(df[col])
```

```python
df.head(10).T
```

```python
df.to_csv("Feature_Engineering_done.csv")
```
