```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('EDA_done.csv')
df.head(10).T
```

```python
print(df.shape)
print(df.columns)
```

```python
df.isnull().sum()
```

```python
df = df.drop(columns=['Blood Pressure (mmHg)'])
```

```python
df = pd.get_dummies(df,columns=["Sex"] , dtype=int)
```

```python
from sklearn.preprocessing import OrdinalEncoder

oe = OrdinalEncoder(categories=[
    ['N', 'Y'],                      # Diabetes Status
    ['N', 'Y'],                      # Smoking Status
    ['LOW', 'INTERMEDIARY', 'HIGH'], # CVD Risk Level
    ['Normal', 'Elevated', 'Hypertension Stage 1', 'Hypertension Stage 2']  # Blood Pressure Category
],dtype=int)

df[['Smoking Status','Diabetes Status', 'CVD Risk Level', 'Blood Pressure Category']] = oe.fit_transform(df[['Smoking Status','Diabetes Status', 'CVD Risk Level', 'Blood Pressure Category']])
```
